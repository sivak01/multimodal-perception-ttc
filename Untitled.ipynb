```python
import subprocess
import sys
from pathlib import Path

def convert_notebooks_to_html(folder_path, output_folder=None):
    """
    Convert all .ipynb files in a folder to HTML.

    Args:
        folder_path: Path to folder containing .ipynb files
        output_folder: Optional path to save HTML files (defaults to same folder)
    """
    folder = Path(folder_path)
    if not folder.is_dir():
        print(f"Error: {folder_path} is not a valid directory")
        return

    notebooks = list(folder.glob("*.ipynb"))

    if not notebooks:
        print(f"No .ipynb files found in {folder_path}")
        return

    print(f"Found {len(notebooks)} notebook(s) to convert.\n")

    out_dir = Path(output_folder) if output_folder else folder
    out_dir.mkdir(parents=True, exist_ok=True)

    for nb in notebooks:
        print(f"Converting: {nb.name} ...")
        try:
            subprocess.run(
                [
                    sys.executable, "-m", "nbconvert",
                    "--to", "html",
                    str(nb),
                    "--output-dir", str(out_dir)
                ],
                check=True,
                capture_output=True,
                text=True
            )
            print(f"  ✔ Saved to: {out_dir / (nb.stem + '.html')}")
        except subprocess.CalledProcessError as e:
            print(f"  ✘ Failed to convert {nb.name}")
            print(f"    {e.stderr}")

    print("\nDone.")

if __name__ == "__main__":
    # Example usage
    folder = input("Enter the path to the folder containing notebooks: ").strip()
    output = input("Enter output folder (press Enter to save alongside notebooks): ").strip()
    convert_notebooks_to_html(folder, output or None)
```
